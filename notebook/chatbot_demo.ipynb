```python

import pandas as pd

questions = {
    "QUESTION": [
        "What is Artificial Intelligence?",
        "What is Machine Learning?",
        "What is Generative AI?",
        "What is Natural Language Processing?",
        "What is Deep Learning?"
    ],
    "CATEGORY": [
        "AI",
        "Machine Learning",
        "Generative AI",
        "NLP",
        "Deep Learning"
    ]
}

df = pd.DataFrame(questions)

csv_path = DATASET_DIR / "sample_questions.csv"
df.to_csv(csv_path, index=False)

print("Sample questions saved successfully!")
print(df)
```

```python

response = ollama.chat(
    model="llama3.2:latest",
    messages=[
        {
            "role": "user",
            "content": "What is Artificial Intelligence? Explain in simple English."
        }
    ]
)

print(response["message"]["content"])
```

```python

answer = response["message"]["content"]

response_data = pd.DataFrame([{
    "QUESTION": "What is Artificial Intelligence?",
    "RESPONSE": answer,
    "TIMESTAMP": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}])

response_path = DATASET_DIR / "responses.csv"
response_data.to_csv(response_path, index=False)

print("Response saved successfully!")
print(response_data[["QUESTION", "RESPONSE"]])
```

```python

import pandas as pd
import ollama
from pathlib import Path
from datetime import datetime

BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebook" else Path.cwd()
DATASET_DIR = BASE_DIR / "dataset"

questions_df = pd.read_csv(DATASET_DIR / "sample_questions.csv")

print("Total questions:", len(questions_df))
```

```python

all_responses = []

for question in questions_df["QUESTION"]:
    print("Processing:", question)

    result = ollama.chat(
        model="llama3.2:latest",
        messages=[
            {"role": "user", "content": question}
        ]
    )

    all_responses.append({
        "QUESTION": question,
        "RESPONSE": result["message"]["content"],
        "TIMESTAMP": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    })

responses_df = pd.DataFrame(all_responses)

responses_df.to_csv(
    DATASET_DIR / "responses.csv",
    index=False
)

print("\nTotal responses saved:", len(responses_df))

```
